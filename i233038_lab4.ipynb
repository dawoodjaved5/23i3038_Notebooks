```python
city_graph = {
    'Depot': [('A', 5), ('B', 2)],
    'A': [('C', 4), ('D', 2)],
    'B': [('A', 8), ('D', 7)],
    'C': [('Customer', 3)],
    'D': [('Customer', 1)],
    'Customer': []
}

initial_state = 'Depot'
goal_state = 'Customer'

def get_actions(state):
    """Possible Actions: MoveTo(Location)"""
    return city_graph.get(state, [])

def transition_model(state, action):
    """RESULT(State, MoveTo(Location))"""
    return action[0]  # Returns the node name

def goal_test(state):
    """Check if current state equals customer location"""
    return state == goal_state

reachable_states = list(city_graph.keys())

def find_path(current_node, visited, path, cost):
    if goal_test(current_node):
        return path, cost
    
    for neighbor, weight in get_actions(current_node):
        if neighbor not in visited:  # Avoid cycles
            new_path, total_cost = find_path(
                neighbor, 
                visited + [neighbor], 
                path + [f"MoveTo({neighbor})"], 
                cost + weight
            )
            if new_path: return new_path, total_cost
            
    return None, 0

print(f"Route Planning Agent")
print(f"Initial State: {initial_state}")
print(f"Goal State: {goal_state}")
print(f"State-Space Graph (Nodes): {reachable_states}")

solution_actions, total_path_cost = find_path(initial_state, [initial_state], [], 0)

if solution_actions:
    print("Solution Action Sequence:")
    for action in solution_actions:
        print(f" {action}")
    print(f"\nTotal Path Cost: {total_path_cost}")
else:
    print("No path found.")
```

```python
goal_state = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

initial_state = (
    (1, 2, 3),
    (4, 0, 7),
    (6, 5, 8)
)

states_generated = 0

def find_blank(state):
    """Locate row and column of the blank space (0)"""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_successors(state):
    """Generate all valid movements for the blank space"""
    r, c = find_blank(state)
    successors = []
    moves = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]
    
    for dr, dc, move_name in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_list = [list(row) for row in state]
            new_list[r][c], new_list[nr][nc] = new_list[nr][nc], new_list[r][c]
            new_state = tuple(tuple(row) for row in new_list)
            successors.append((new_state, move_name))
            
    return successors

def solve(current_state, depth, visited, path):
    global states_generated
    states_generated += 1
    
    if current_state == goal_state:
        return path
    
    if depth <= 0:
        return None
    
    for next_state, move in get_successors(current_state):
        if next_state not in visited:
            visited.add(next_state)
            result = solve(next_state, depth - 1, visited, path + [move])
            if result: return result
            visited.remove(next_state) # Backtrack
            
    return None

print("8-puzzle solver")
print(f"Initial State: {initial_state}")
print(f"Goal State: {goal_state}\n")

depth_limit = 10
solution = solve(initial_state, depth_limit, {initial_state}, [])

print(f"Total States Generated: {states_generated}")
if solution:
    print(f"Solution Path: {' -> '.join(solution)}")
    print(f"Total Path Cost (Moves): {len(solution)}")
else:
    print(f"No solution found within depth limit of {depth_limit}")
```
